## Data Types : 

### str :

#### Multiple ways to create a string in Python

```python
a = "Kathir"
b = 'kavi'
c = str("Kavi") # Setting a string value to the variable 'c' using the str() constructor
d : str = "John"  # Assigning a string value to the variable 'd' with type hinting
e = """Hey,
how are you?""" # Preserving line breaks in a string using triple quotes
f = '''Hey,
Hello Nice to meet you''' # Preserving line breaks in a string using triple quotes
g = "Hey,\n Hello Nice to meet you" # Preserving line breaks in a string using escape characters
h = "It\'s Python" # \' represents a single quote inside a string.
i = r"Hey,\n Hello Nice to meet you" # r represents a raw string where escape characters are not processed
j = f"Hey, myself {a} !" # f represents a formatted string where variables can be embedded directly into the string



print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
print(h)
print(i)
print(j)


k = 24
print(k)
print(type(k))
l = str(k) # Converting an integer to a string using the str() constructor
print(l)
print(type(l))

m = ("Heyy" "Hi") # Concatenating two strings using parentheses
n = "Heyy" + "Hi" # Concatenating two strings using the + operator
o = "Heyy" * 3 # Repeating a string multiple times using the * operator
print(m)
print(n)
print(o)

p = str() # Creating an empty string using the str() constructor
print(p) # returns an empty string
```

### int :

#### Multiple ways to create a integer in Python

```python
a = 23
b : int = 24 # Assigning an integer value to the variable 'b' with type hinting
c = "25"
d = int(c) # Converting a string to an integer using the int() constructor
e = 23.5
f = int(e) # Converting a float to an integer using the int() constructor
g = 1_000_000 # Using underscores for better readability in large numbers
h = int() # Creating an integer with a default value of 0 using the int() constructor


print(a)
print(type(a))
print(b)
print(type(b))
print(c)
print(type(c))
print(d)
print(type(d))
print(e)
print(type(e))
print(f)
print(type(f))
print(g)
print(type(g))
print(h)
```

### float :

#### Multiple ways to create a float in Python

```python
a = 2.4
b = 3
c = float(b) # Converting an integer to a float using the float() constructor
d : float = 4.5 # Assigning a float value to the variable 'd' with type hinting
e = float() # Creating a float with a default value of 0.0 using the float() constructor

print(a)
print(type(a))
print(b)
print(type(b))
print(c)
print(type(c))
print(d)
print(type(d))
print(e)
```

### range :

#### Multiple ways to create a range in Python


```python
# range(start: Optional[int], stop: int, step: Optional[int])

a = range(5) # Creating a range object that represents a sequence of numbers from 0 to 4
b = range(1, 10) # Creating a range object that represents a sequence of numbers from 1 to 9
c = range(1, 10, 2) # Creating a range object that represents a sequence of odd numbers from 1 to 9 
d = range(1,0) # Creating a range object that represents an empty sequence since the start is greater than the stop
e = range(10, 0, -1) # Creating a range object that represents a sequence of numbers from 10 to 1 in reverse order
f = range(2,2) # Creating a range object that represents an empty sequence since the start and stop are the same
g = range(-3, 2) # Creating a range object that represents a sequence of numbers from -3 to 1

print(a)
print(list(a)) # Converting the range object to a list to see the actual numbers
print(b)
print(list(b))
print(c)
print(list(c))
print(d)
print(list(d)) # This will return an empty list since the range is invalid
print(e)
print(list(e))
print(f)
print(list(f)) # This will return an empty list
print(g)
print(list(g))
```

### bool :

#### Multiple ways to create a bool in Python


```python
a = True # Assigning a boolean value to the variable 'a'
b : bool = False # Assigning a boolean value to the variable 'b' with type hinting
c = bool() # Creating a boolean with a default value of False using the bool() constructor
d = bool(1) # Creating a boolean with a value of True since 1 is considered true
e = bool(0) # Creating a boolean with a value of False since 0 is considered false
f = bool("Hello") # Creating a boolean with a value of True since non-empty strings are considered true
g = bool("") # Creating a boolean with a value of False since an empty string is considered false
h = bool([]) # Creating a boolean with a value of False since an empty list is considered false
i = bool([1, 2, 3]) # Creating a boolean with a value of True since non-empty lists are considered true
j = bool(None) # Creating a boolean with a value of False since None is considered false
k = bool({}) # Creating a boolean with a value of False since an empty dictionary is considered false
l = bool({"key": "value"}) # Creating a boolean with a value of True since non-empty dictionaries are considered true

print(a)
print(type(a))
print(b)
print(type(b))
print(c)
print(type(c))
print(d)
print(type(d))
print(e)
print(type(e))
print(f)
print(type(f))
print(g)
print(type(g))
print(h)
print(type(h))
print(i)
print(type(i))
print(j)
print(type(j))
print(k)
print(type(k))
print(l)
print(type(l))
```

### list :

#### Mutiple ways to create a list in Python 

```python
a = [5,1,2,3,4] # Creating a list with multiple elements
b : list[int] = [1,2,3,4,5] # Creating a list with multiple elements and type hinting
c = list() # Creating an empty list using the list() constructor
d = [1, "Hello", 3.14, True] # Creating a list with mixed data types
e = ["Hello"] # Creating a list with a single string element
f = list(range(5)) # Creating a list from a range object
g = list("Hello") # Creating a list from a string, where each character becomes an element in the list
h = [] # Creating an empty list using square brackets
i = [[1,2],[3,4]] # Creating a nested list (a list of lists)
j = [1,2] + [3,4] # Concatenating two lists using the + operator
k = [1,2] * 2 # Repeating a list using the * operator
l = [0 for i in range(5)] # Creating a list using a list comprehension that generates a list of zeros
m = [i for i in range(5)] # Creating a list using a list comprehension that generates a list of numbers from 0 to 4

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
print(h)
print(i)
print(j)
print(k)
print(l)
print(m)
```
